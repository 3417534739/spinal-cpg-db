## Abstract

> Amphibious fishes moving from water to land experience continuous changes in environmental forces. How these subtle changes impact behavioural transitions cannot be resolved by comparisons of aquatic and terrestrial locomotion. For example, aquatic and terrestrial locomotion appear distinct in the actinopterygian fish Polypterus senegalus ; however, it is unclear how gradual water level changes influence the transition between these locomotor behaviours. We tested the hypothesis in P. senegalus that swimming and walking are part of an incremental continuum of behaviour and muscle activity across the environmental transition from water to land rather than two discrete behaviours, as proposed by previous literature. We exposed P. senegalus to discrete environments from fully aquatic to fully terrestrial while recording body and pectoral fin kinematics and muscle activity. Anterior axial red muscle effort increases as water depth decreases; however, a typical swimming-like anterior-to-posterior wave of axial red muscle activity is always present, even during terrestrial locomotion, indicating gradual motor control changes. Thus, walking appears to be based on swimming-like axial muscle activity whereas kinematic differences between swimming and walking appear to be due to mechanical constraints. A discrete change in left–right pectoral fin coordination from in-phase to out-of-phase at 0.7 body depths relies on adductor muscle activity with a similar duty factor and adductor muscle effort that increases gradually as water depth decreases. Thus, despite distinct changes in kinematic timing, neuromuscular patterning is similar across the water depth continuum.……（后略）

**中文翻译**

从水中移动到陆地的两栖鱼经历环境力的连续变化。这些细微变化如何影响行为转换，无法通过水生与陆生运动的简单比较来解决。例如，辐鳍鱼 Polypterus senegalus 的水生与陆生运动看似截然不同，但水位逐渐变化如何影响这两种运动行为之间的转换尚不清楚。我们在 P. senegalus 中检验以下假说：游泳与行走是跨环境从水到陆过渡中行为与肌肉活动的渐进连续体，而非以往文献提出的两种离散行为。我们将 P. senegalus 置于从完全水生到完全陆生的离散环境中，同时记录身体与胸鳍的运动学及肌肉活动。前部轴肌红肌的用力随水深降低而增加；然而，典型的游泳式前-后轴肌红肌活动波始终存在，即使在陆上运动时也是如此，提示运动控制的渐进变化。因此，行走似乎基于游泳样的轴肌活动，而游泳与行走的运动学差异似乎源于机械约束。左右胸鳍协调在 0.7 体深处从同相到反相的离散变化依赖于内收肌活动，其占空比相似，内收肌用力随水深降低而逐渐增加。因此，尽管运动学时序明显变化，跨水深连续体的神经肌肉模式是相似的。
