## Abstract

> Ciona intestinalis Neuronal Calcium Imaging Dataset. Contributed by: Kushal Kolar, Daniel Dondorp, Marios Chatzigeorgiou. This dataset (CRCNS ci-1) contains raw GCaMP6s calcium imaging recordings of multiple genetically defined groups of neurons in the Ciona larval CNS, plus epidermal CesA+ and Trunk Lateral HNK-1 cell groups; the data are hosted on Figshare (doi 10.6084/m9.figshare.10289162) in Mesmerize FAIR project format (~477.6 GB uncompressed) and can fully reproduce the published analysis.
