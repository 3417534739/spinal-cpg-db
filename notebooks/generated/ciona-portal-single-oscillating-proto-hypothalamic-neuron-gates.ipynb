## Abstract

> Summary Ciona larvae display a number of behaviors, including negative phototaxis. In negative phototaxis, the larvae first perform short spontaneous rhythmic casting swims. As larvae cast in a light field, their photoreceptors are directionally shaded by an associated pigment cell, providing a phototactic cue. This then evokes an extended negative taxis swim. We report here that the larval forebrain of Ciona has a previously uncharacterized single slow-oscillating inhibitory neuron (neuron cor-assBVIN78 ) that projects to the midbrain, where it targets key interneurons of the phototaxis circuit known as the photoreceptor relay neurons . The anatomical location, gene expression and oscillation of cor-assBVIN78 suggest homology to oscillating neurons of the vertebrate hypothalamus. Ablation of cor-assBVIN78 results in larvae showing extended phototaxis-like swims, but which occur in the absence of phototactic cues. These results indicate that cor-assBVIN78 has a gating activity on phototaxis by projecting temporally-oscillating inhibition to the photoreceptor relay neurons. However, in intact larvae the frequency of cor-assBVIN78 oscillation does not match that of the rhythmic spontaneous swims, indicating that the troughs in oscillations do not themselves initiate swims, but rather that cor-assBVIN78 may modulate the phototaxis circuit by filtering out low level inputs while restricting them temporally to the troughs in inhibition.

Ciona 幼体表现出多种行为，包括负趋光。在负趋光中，幼体先进行短促的自发节律性 casting 游泳。当幼体在光场中摆动时，其光感受器被伴行色素细胞方向性遮挡，提供趋光线索，随后引发延长的负趋光游泳。我们报道 Ciona 幼体前脑存在一个此前未鉴定的单个缓慢振荡抑制神经元（cor-assBVIN78），它投射到中脑，靶向趋光回路的关键中间神经元——光感受器中继神经元。cor-assBVIN78 的解剖位置、基因表达与振荡特征提示其与脊椎动物下丘脑的振荡神经元同源。消融 cor-assBVIN78 后，幼体表现出延长的趋光样游泳，但发生在没有趋光线索的情况下。这些结果表明 cor-assBVIN78 通过向光感受器中继神经元投射时间振荡的抑制，对趋光行为起门控作用。然而在完整幼体中，cor-assBVIN78 的振荡频率与节律性自发游泳的频率并不匹配，说明振荡低谷本身并不直接发起游泳，而是 cor-assBVIN78 通过过滤低水平输入并将其时间性地限制在抑制低谷内来调制趋光回路。
