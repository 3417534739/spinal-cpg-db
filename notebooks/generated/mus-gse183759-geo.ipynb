## Abstract（原文摘抄 + 中文翻译）**原文（EN）**：> Spinal motor neurons (MNs) integrate sensory stimuli and brain commands to generate movements. In vertebrates, the molecular identities of the cardinal MN types such as those innervating limb versus trunk muscles are well elucidated. Yet the identities of finer subtypes within these cell populations that innervate individual muscle groups remain enigmatic. Here we investigate heterogeneity in mouse MNs using single-cell transcriptomics. Among limb-innervating MNs, we reveal a diverse neuropeptide code for delineating putative motor pool identities. Additionally, we uncover that axial MNs are subdivided into three molecularly distinct subtypes, defined by mediolaterally-biased Satb2, Nr2f2 or Bcl11b expression patterns with different axon guidance signatures. These three subtypes are present in chicken and human embryos, suggesting a conserved axial MN expression pattern across higher vertebrates. Overall, our study provides a molecular resource of spinal MN types and paves the way towards deciphering how neuronal subtypes evolved to accommodate vertebrate motor behaviors. Subject terms: Development of the nervous system, Transcriptomics, Cellular neuroscience How molecular diversity in neurons links to versatile functions is elusive. Here the authors profiled embryonic spinal motor neurons with single-cell RNAseq and identified molecular subtypes targeting distinct muscle groups in different species.**中文翻译**：脊髓运动神经元（MN）整合感觉刺激与脑指令以产生运动。在脊椎动物中，支配肢体与躯干肌肉等主要 MN 类型的分子身份已被清楚阐明。然而这些细胞群内支配各肌肉群的更细亚型的身份仍是谜。我们用单细胞转录组学研究小鼠 MN 的异质性。在支配肢体的 MN 中，我们揭示了界定假定运动池身份的多样神经肽编码。此外，我们发现轴向 MN 分为三个分子上不同的亚型，由内外侧偏向的 Satb2、Nr2f2 或 Bcl11b 表达模式定义，并带有不同的轴突导向特征。这三个亚型存在于鸡胚与人胚中，提示高等脊椎动物中保守的轴向 MN 表达模式。总体而言，我们的研究提供了脊髓 MN 类型的分子资源，为解读神经元亚型如何演化以适应脊椎动物运动行为铺平道路。## Highlights- Mnx1-GFP 富集（代码 github.com/sqjin/scRNA-seq-motor-neuron）## Data description (数据介绍)- **Source**: portal/repository page — https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE183759- **From the entry record**: Mnx1-GFP 富集（代码 github.com/sqjin/scRNA-seq-motor-neuron）- 取材详情（物种/性别/组织/时期/数量）：详见数据页
